```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

df = pd.read_csv("../data/processed/data_initial_audit.csv")

print("Dataset shape:", df.shape)
display(df.head())
```

```python
# Target variable
y = df["event"].astype(int)

# Columns that must not be used as prediction features
excluded_columns = [
    "customer_id",
    "starting_date",
    "suspension_date",
    "observation_end_date",
    "duration_days",
    "duration_years",
    "event"
]

# Features
X = df.drop(columns=excluded_columns)

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

display(X.head())
```

```python
print("Feature column names:")
print(X.columns.tolist())

print("\nMissing values:")
display(X.isnull().sum().sort_values(ascending=False))

print("\nData types:")
display(X.dtypes)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)

print("\nTraining event distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting event distribution:")
print(y_test.value_counts(normalize=True))
```
