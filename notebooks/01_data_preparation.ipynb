# Insurance Lapse & CLV Prediction

## Step 1: Data Loading and Initial Inspection

Objectives:
1. Load the dataset
2. Inspect the shape and columns
3. Understand data types
4. Check missing values
5. Identify potential data quality issues

```python
import pandas as pd
import numpy as np
import os
file_path = "../data/processed/eda_dataset.csv"

print("Current working directory:", os.getcwd())
print("File exists:", os.path.exists(file_path))

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
# Convert date columns into datetime format

date_columns = [
    "starting_date",
    "suspension_date",
    "observation_end_date"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors="coerce")

print("Date conversion completed.")

display(df[date_columns].head())
```

```python
# Basic validation

print("Shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes)

print("\nMissing values:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
display(df.dtypes)
```

```python
missing_values = df.isnull().sum().sort_values(ascending=False)

print("Missing values in each column:")
display(missing_values)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
display(df.describe(include="all").T)
```

```python
# Check survival-related columns

print("Event value counts:")
display(df["event"].value_counts(dropna=False))

print("\nDuration days summary:")
display(df["duration_days"].describe())

print("\nDuration years summary:")
display(df["duration_years"].describe())

print("\nEvent and duration missing values:")
display(df[["event", "duration_days", "duration_years"]].isnull().sum())
```

```python
# Check for invalid survival values

print("Event values:")
print(sorted(df["event"].dropna().unique()))

print("\nRows with negative duration:")
print((df["duration_days"] < 0).sum())

print("\nRows with zero or negative annual premium:")
print((df["annual_premium"] <= 0).sum())

print("\nRows with invalid age:")
print((df["age"] <= 0).sum())
```

```python
# Save a copy after date conversion

output_path = "../data/processed/data_initial_audit.csv"

df.to_csv(output_path, index=False)

print("Initial audit dataset saved:", output_path)
```
